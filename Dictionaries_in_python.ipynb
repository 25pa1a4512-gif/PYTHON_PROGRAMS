***WRITE A PROGRAM TO CHECK THE GIVEN NUMBER EXISTS IN DICTIONARY OR NOT***

```python
#CREATE A DICTIONARY
student={
    "name":"jessy",
    "age":"19",
    "course":"ai'ds"
}
key=input("enter the key to search: ")
if key in student:
  print("key exist in dictionary")
else:
  print("key does not exist in tje dictionary")
```

***WRITE A PROGRAM TO ADD A KEY-VALUE PAIR TO AN EXISTIND DICTIONARY***

```python
d={'a':1,'b':2,'c':3,'d':4,'e':5}
key=input("Enter a key:")
value=input("enter a value:")
d[key]=value
print(d)
```

***WRITE A PROGRAM TO SUM ALL THE ITEMS IN A GIVEN DICTIONARY***

```python
d={'a':1,'b':2,'c':3,'d':4,'e':5}
total=sum(d.values())
print(total)
```

***WRITE A PROGRAM THAT PRINTS FREQUENCY OR OCCURANCE OF EACH ELEMENT IN A LIST IN FOLLOWING***
***FORMAT[1,2,3,4,1,3,5,5,6,6,3,2,2,3] {1:2,2:3,3:3,4:1,5:2,6:2}***

```python
L=[int(i) for i in input().split()]
d={}
for i in L:
  d[i]=L.count(i)
print(d)
```
