```python
#List operations
#1.concatination
#2.repetition
#3.indexing
#4.slicing
#5.membership operators
#6.iteration

#1.cocatination
L1=(1,2,3,"hello")
L2=(22,33,44)
print(L1+L2)

#2.repetition
print(L1*3)

#3.indexing
print(L1[3])
print(L1[-4])

#4.slicing
print(L1[0:3])
print(L1[1:])
print(L1[-4])
L2=[1,2,3,4,5,6,77,88,99,100]
print(L2[0:5:2])
print(L2[-4:-1])
print(L2[::-1])
print(L2[::-2])

#membership
if 100 in L2:
  print("yes")
if 10 not in L2:
  print("yes")

#iteration
for i in L2:
  print(i)
```

```python
#List with sequence
L=[]
for i in range (4):
  L.append(i**2)
print(L)
#L=[value for i in sequence]
#L=[i**2 for i in range(4)]
#print(L)
```

```python
# List comprehension
L2=[i**3 for i in L]
print(L2)
```
