***SET OPERATIONS***

```python
#UNION
#INTERSECTION
#DIFFERENCE
#SYMMETRIC DIFFERENCE

#union
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.union(B)

print("Set A:", A)
print("Set B:", B)
print("Union:", result)

```

```python
#intersection
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.intersection(B)

print("Set A:", A)
print("Set B:", B)
print("Intersection:", result)
```

```python
#difference
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.difference(B)

print("Set A:", A)
print("Set B:", B)
print("Difference A - B:", result)
```

```python
#symmetric difference
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.symmetric_difference(B)

print("Set A:", A)
print("Set B:", B)
print("Symmetric Difference:", result)
```

***SET METHODS***

```python
 #add
 #update
 #remove
 #dicard
 #pop
 #clear
 #union
 #intersection
 #difference
 #symmetric difference
```

```python
#add() Adds one element to a set.

fruits = {"apple", "banana", "mango"}

fruits.add("orange")

print(fruits)

```

```python
#update() Adds multiple elements to a set.

fruits = {"apple", "banana"}

fruits.update(["mango", "orange", "grapes"])

print(fruits)
```

```python
#remove() Removes a specified element. If the element does not exist, remove() gives a KeyError.

numbers = {10, 20, 30, 40}

numbers.remove(30)

print(numbers)
```

```python
#discard() Also removes an element, but unlike remove(), it doesn't give an error if the element doesn't exist.

numbers = {10, 20, 30, 40}

numbers.discard(30)

print(numbers)
```

```python
#pop() Removes and returns one arbitrary element from the set.

numbers = {10, 20, 30, 40}

x = numbers.pop()

print("Removed:", x)
print("Set:", numbers)
```

```python
#clear() Removes all elements from the set.

numbers = {10, 20, 30, 40}

numbers.clear()

print(numbers)
```

```python
#union() Combines all elements from both sets.

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

result = A.union(B)

print(result)
```

```python
#intersection() Returns elements that are common to both sets.

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

result = A.intersection(B)

print(result)
```

```python
#difference() Returns elements that are in the first set but not in the second set.

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

result = A.difference(B)

print(result)
```

```python
#symmetric_difference() Returns elements that are in either set, but NOT in both.

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

result = A.symmetric_difference(B)

print(result)
```

```python
#program to find no.of duplicates
numbers = [1, 2, 3, 2, 4, 5, 1, 6, 3]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate elements:", duplicates)
print("Number of duplicates:", len(duplicates))
```
