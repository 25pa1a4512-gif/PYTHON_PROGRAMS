```python
with open("f1.txt", "a") as f:
    f.writelines(["\napple", "\nname", "\nend"])

with open("f1.txt", "r") as f:
    print(f.readlines())
```

***Or, if you want to demonstrate read(), readlines(), etc.:***

```python
with open("f1.txt", "a") as f:
    f.writelines(["\napple", "\nname", "\nend"])

with open("f1.txt", "r") as f:
    data = f.readlines()
    print(data)
```

***If you want to read the file multiple times, use f.seek(0) to move the file pointer back to the beginning:***

```python
with open("f1.txt", "r") as f:
    print(f.readlines())

    f.seek(0)
    print(f.read())

    f.seek(0)
    data = f.readlines()
    print(data)
```

```python
#copying content of one file to another
with open("f1.txt","r")as f:
  data=f.read()
  print("reading done")
  with open("f2.txt","w") as f2:
    f2.write(data)
    print("file copied")
```

```python
#print each line of a file in reverse order
with open("new.txt","r")as file:
  data=file.readlines()
  for line in data[::-1]:
    print(line)
```

```python
#python program to compute the number of characters, words and lines in a file
with open("f1.txt", "r") as f:
    data = f.read()

characters = len(data)
words = len(data.split())
lines = len(data.splitlines())

print("Number of characters:", characters)
print("Number of words:", words)
print("Number of lines:", lines)
```
